# Конспекты по «Цифровой связи»

Представлено конспективное изложение основ «Цифровой связи» — общепрофессиональной дисциплины, которая может называться по-разному:

* «Общая теория связи»,
* «Передача дискретных сообщений»,
* «Теория и техника передачи информации»,
* «Формирование и обработка сигналов систем связи» и т.п.

Цель настоящих конспектов — изложение материала в достаточно общей форме без особенного погружения в детали, что даст изучающему фундамент, который позволит изучать более детальные и, как правило, менее доступные для понимания книги, статьи или даже учебные пособия.

Данное изложение является некой мозаикой, которая сложилась у автора после 9-летнего опыта преподавания данной дисциплины в техническом вузе.

Материал структурирован в виде небольшого количества абзацев, разделённых подзаголовками, что позволяет быстро отыскивать материал в случае надобности, а также сосредотачиваться на так называемых «моментах» и осмысливать их дольше.

Для освоения данных конспектов необходимы знания:

* высшей математики и
* теории вероятностей в объёме технического вуза, а также
* школьной физики.

Будет плюсом знание дисциплин «Радиотехнические цепи и сигналы», «Основы теории цепей» или подобных.

## Цифровая связь как транспортный сервис. Условие успешного приёма.

Под цифровой связью подразумевают техническую передачу символов с помощью некоторых сигналов через определённые каналы связи.

Символами называют элементы заранее определённого алфавита. Каждому символу — двоичному слову фиксированной длины — ставится в соответствие сигнал определённой формы, информационный импульс. Например, таблица соответствия может быть следующей:

$$(0, 0, 0) \rightarrow s_0(t),$$
$$(0, 0, 1) \rightarrow s_1(t),$$
$$(0, 1, 0) \rightarrow s_2(t),$$
$$\ldots$$
$$(1, 1, 1) \rightarrow s_7(t).$$

Для того чтобы приём был принципиально возможен, сигналы $s_i(t)$ должны отличаться друг от друга, т.е. если $i \neq j$, то и $s_i(t) \neq s_j(t)$. Как видим, каких-то особенно сильных ограничений к сигналам не предъявляется; они могут в том числе перекрываться по времени.

В итоге имеем набор пронумерованных сигналов, которые предназначены для передачи цифровой информации (двоичных кодов — символов) по линии связи. Именно это формальное требование подчёркивает техническую сторону цифровой связи как некоторого транспортного сервиса, которому безразлично содержание передаваемой информации.

## Канал связи. Качество канала связи. Надёжная связь.

Под каналом связи понимают согласованный набор технических средств, который позволяет, задавая определённое входное воздействие, наблюдать связанный с ним отклик.

Чем безошибочнее мы можем судить о входном воздействии по наблюдаемому отклику, тем лучше качество канала связи.

Для поддержания заданного качества тратится некоторая энергия. Чем выше требуемое качество, тем больше тратится энергии на передачу и приём одного бита (бит — двоичный символ), однако вся соль в том, что степень роста затрачиваемой энергии зависит от конкретного канала связи. Стремятся, чтобы меры по улучшению качества канала не приводили к резкому росту потребляемой энергии.

В цифровых системах связи качество канала связи определяется вероятностью ошибки принятого символа. Чем качественнее канал, тем меньше повторных запросов потребуется для обеспечения надёжной связи, при этом под надёжной связью понимают практически идеальную передачу информации, когда итоговой вероятностью ошибки можно пренебречь в рамках решаемой задачи. Современные мобильные системы связи при хороших условиях приёма могут обеспечивать вероятность битовой ошибки от ${10}^{-8}$ до ${10}^{-6}$ (это всего лишь ориентир, т.к. системы связи продолжают развиваться).

## Линия связи.

Под линией связи понимают некоторую физическую среду передачи, которая направляет сигнал, несущий передаваемую информацию.

Направление сигнала необходимо как для экономии энергии, так и для разграничения каналов, чтобы не мешать параллельно работающим системам связи и абонентам. Примерами направляющих сред могут быть любые кабели (телефонный кабель типа «лапша», витая пара, коаксиальный кабель, оптическое волокно), СВЧ-линии, волноводы, водная и воздушная среды. В случае водной и воздушной сред направленность сама по себе слаба, однако она создаётся антенными системами, которые можно отнести к направляющим средам.

Линия связи входит в канал связи (вложена в него) и не может быть из него исключена принципиально. Канал связи — понятие логическое и растяжимое; он может включать в себя (помимо линии связи) модемы, скремблеры, кодеки, шифраторы и т.п. Канал связи (его граница) задаётся в зависимости от текущей решаемой задачи: текущего уровня анализа или синтеза.

## Непрерывность сигнала. Дискретность цифровой информации.

Под сигналом понимают некоторую величину, которая изменяется с течением времени. В системах связи часто это напряжение или ток в электрической цепи, интенсивность света в оптическом волокне, комплексная амплитуда электромагнитного поля в СВЧ-линии. Этой изменяющейся величине соответствует некоторая информация.

В системах цифровой связи информация является дискретной по своей природе, т.е. счётной; сам же сигнал всегда непрерывен безотносительно к типу рассматриваемой системы.

В аналоговых системах связи информацией является непрерывный сигнал, полученный после всех преобразований в приёмнике. Неотъемлемой частью такого сигнала является шум, который можно снижать, но нельзя устранить принципиально. В цифровых же системах связи принятая информация в большинстве случаев не содержит ошибок, т.е. приёмник располагает исходными значениями битов — нулей и единиц.

Следует отметить, что теоретическое рассмотрение сигналов подразумевает их деление на дискретные и непрерывные (причём как по времени, так и по уровню), однако в рамках дисциплины «Цифровая связь» достаточно рассматривать сигналы только как непрерывные, что противопоставляет их дискретной по своей природе цифровой информации. Было бы глупо рассматривать цифровую информацию как некоторую непрерывную величину, но, с другой стороны, её передача и есть цифровая связь как таковая, потому что нельзя передать информацию, не превратив её в сигнал, пригодный для передачи по выбранной линии связи.

Превращение информации в сигнал является примером насыщения информации энергией, что подчёркивает дуализм «энергия — информация» и, вероятно, может служить аналогией известному в физике дуализму «волна — частица».

## Энергия на один бит информации. Битовая скорость передачи информации.

Как теоретически, так и практически в простейшем случае требуется передавать биты — двоичные символы. На передачу любой информации тратится некоторое время. Чем меньше тратится времени, тем выше скорость передачи информации.

Сигналы цифровых систем связи характеризуются средней мощностью, которая вкупе с затраченным временем определяет затраченную энергию. Для целей сравнения разных систем связи удобно рассматривать энергию на один бит информации:

$$E_b = P T_b.$$

Здесь $P$ — средняя мощность сигнала, $T_b$ — интервал следования битов (в секундах). При этом битовая скорость передачи информации будет равна:

$$V_b = 1 / T_b, ~\text{бит/с}.$$

Качество цифровых систем передачи информации напрямую зависит от энергии на один бит. Поясним, почему это так.

## Почему качество цифровой связи определяется энергией, а не мощностью? Помехоустойчивое кодирование.

Само собой разумеется, что на качество связи влияет уровень шума и помех, однако на данном этапе можно полагать, что этот уровень постоянен.

Средняя мощность сигнала определяет «громкость» сигнала, но качество связи зависит не только от неё. Длительность бита также влияет, потому что чем дольше происходит наблюдение некоторого сигнала, кодирующего один бит, тем больше информации о сигнале накапливается и тем выше вероятность правильного решения относительно переданного бита. Это тем актуальнее, чем ниже отношение сигнал-шум, т.е. хуже условия приёма. При отличных условиях приёма длительность бита можно уменьшить, сохранив то же качество связи, при этом повысится скорость передачи информации; справедливо и обратное: при ухудшении условий приёма увеличивают длительность информационного бита, чтобы сохранить требуемое качество связи, однако скорость передачи при этом уменьшается. Для заданной скорости передачи существует предельное отношение сигнал-шум, ниже которого надёжная связь невозможна. Увеличение длительности бита снижает это требование, но уменьшает скорость; абсолютный предел задаётся предельным значением $E_b/N_0$ (предел Шеннона).

Заметим, что изменение длительности информационного бита можно реализовывать по-разному:

1. Физически изменять интервал следования битов,

2. Передавать информационный бит несколькими битами с исходным интервалом следования битов.

В первом случае изменяется ширина спектра сигнала. Во втором случае происходит избыточное кодирование информации (помехоустойчивое кодирование), когда $k$ исходных битов передают с помощью $n > k$ битов. На практике чаще применяют второй способ поддержания заданного качества связи.

## Интенсивный и экстенсивный пути улучшения качества цифровой связи.

Так как качество цифровых систем связи напрямую зависит от энергии на один бит, то намечаются два пути улучшения качества связи:

1. Увеличение средней мощности сигнала,

2. Уменьшение скорости передачи информации (увеличение времени передачи информационного бита).

Первый путь является экстенсивным в том плане, что он эквивалентен, например, расширению площади засеваемых земель для повышения объёма заготавливаемой продукции.

Очевидно, должен существовать интенсивный путь, который в плане сельского хозяйства эквивалентен повышению урожайности культуры за счёт разного рода селекции (генетических манипуляций). Таким путём улучшения качества связи является использование помехоустойчивых кодов, которые, конечно, небесплатны, однако затраты в большей степени связаны с усложнением процедуры кодирования информации и сигнального представления. Это приводит к определённой задержке при обработке информации, а также увеличению потребляемой системой мощности, хотя это увеличение относительно небольшое, если сравнивать его с мощностными затратами экстенсивного пути.

Второй путь, если его понимать как изменение интервала следования битов, также экстенсивен, потому что время — это такой же количественный ресурс, как и средняя мощность сигнала. Если второй путь понимать как помехоустойчивое кодирование, то он будет интенсивным.

Несомненно, интенсивным путём будет улучшение общих технологий передачи и приёма сигналов. Сюда относятся:

* Уменьшение уровня шума приёмников (транзисторные и другие технологии),
* Улучшение характеристик антенных систем (технологии материалов и поверхностей),
* Повышение стабильности генерируемых колебаний (технологии резонаторов),
* Повышение КПД передатчиков сигналов (транзисторные технологии),
* Улучшение экранирования подсистем (технологии материалов и поверхностей).

Однако с точки зрения общей теории связи все улучшения, в итоге, сводятся к:

* Уменьшению вероятности ошибки приёма информационного бита,
* Уменьшению затрат энергии для передачи и приёма информационного бита,
* Увеличению скорости передачи и приёма информационного бита.

Следует отметить, что в системах связи применяются все пути, независимо от того, экстенсивны они или интенсивны. Это связано с требованием адаптации систем связи, потому что каналы связи могут быть достаточно изменчивы по своей природе (например, радиоканалы), а также есть потребность в регулировании параметров по разным протокольным причинам, которые не зависят от качества канала связи — это разного рода регуляторные ограничения, например на уровень излучения или ширину канала.

С точки зрения системы «в целом», введение экстенсивных мер сглаживается наличием интенсивных: последние фактически обеспечивают поддержание приёмлемой эффективности системы при любых допустимых регулировках. Именно поэтому любые интенсивные меры — коды, алгоритмы, сигналы, технологии — очень ценны.

## Техническая и битовая скорости передачи. Символ модуляции. Порядок модуляции. Скорость кодирования.

В цифровых системах связи следует различать две скорости передачи информации — техническую и битовую.

Техническая скорость передачи информации определяется интервалом следования символов $T$; этот интервал напрямую задаёт длительность импульса, и неважно, что данный импульс переносит: служебный бит, информационный бит или целое кодовое слово — символ модуляции (несколько бит):

$$V_T = 1/T, ~\text{бод}.$$

Техническая скорость измеряется в бодах. Именно она главным образом определяет ширину канала (в герцах), которая требуется для передачи данного сигнала. Например, интервалу следования символов 1 мс (1000 символов в секунду) соответствует скорость 1000 бод или 1 кбод, при этом ширина канала должна быть порядка 1 кГц. Точное определение ширины канала возможно только при конкретизации системы связи, когда чётко определён её состав.

Использование помехоустойчивых кодов приводит к уменьшению скорости передачи информации, потому что помимо информационных требуется успевать передавать проверочные биты.

Помехоустойчивые коды характеризуются технической избыточностью (Redundancy): отношением количества проверочных битов $r$ к общей длине кодового слова $n$:

$${R}_{\text{ex}} = r / n, ~~0 \leq {R}_{\text{ex}} < 1.$$

Битовая скорость — это скорость передачи информационных битов, не считая разного рода служебных битов, поэтому в текущем контексте она может быть определена так:

$$V_b = 1 / T_b = (1 - {R}_{\text{ex}}) V_T.$$

Нулевой избыточности соответствует отсутствие помехоустойчивого кодирования, поэтому битовая скорость совпадает с технической (считаем при этом, что никаких других служебных битов нет). Избыточности $0{,}5$ соответствует ситуация, когда один информационный бит передаётся двумя битами, т.е. битовая скорость в два раза ниже технической.

Чтобы не было резкого уменьшения битовой скорости от введения кодирования, вводят модуляцию, когда один символ модуляции представляет слово из нескольких битов. В этом случае выражение для битовой скорости следует уточнить:

$$V_b = (1 - {R}_{\text{ex}}) V_T \log_2 M = R V_T \log_2 M, ~\text{бит/с}.$$

Здесь $M$ — количество символов модуляции (порядок модуляции, размер алфавита). В случае двоичной модуляции символов два, $M = 2$, и один символ переносит ровно один бит. В случае четверичной модуляции символов четыре, $M = 4$, и один символ переносит два бита, и т.д. Величина

$$R = 1 - {R}_{\text{ex}} = 1 - r / n = (n - r) / n = k / n$$
— скорость кодирования (Code Rate). Дальше будем пользоваться именно этой величиной, т.к. она более распространена в формулах.

## О размерности энергии. Децибел. Спектральная плотность мощности шума.

Отметим, что хотя энергия $E$ и измеряется в джоулях как абсолютная физическая величина, на практике в цифровой связи используют безразмерную амплитуду сигналов, привязанную к безразмерным кодам символов модуляции, поэтому остаётся привязка только к секунде, и следовало бы приписывать энергии размерность $[\text{с}]$, однако это сбивало бы с толку, поэтому пишут $[\text{В}^2 \cdot \text{с}]$, где $\text{В}$ — условный «вольт», который фактически безразмерен.

При сравнении систем связи друг с другом строят графики зависимости вероятности битовой ошибки от отношения сигнал-шум (ОСШ), при этом под отношением сигнал-шум понимают безразмерную величину:

$$q^2 = E_b / N_0,$$

которая равна отношению средней энергии на один бит $E_b$ к односторонней спектральной плотности мощности (СПМ) шума $N_0$. Далее везде используется именно односторонняя СПМ, заданная для неотрицательных частот. Именно поэтому в формулах с $N_0$ появляется множитель 2. Зачастую ОСШ — энергетическую величину — удобно выражать в децибелах:

$${q}_{\text{дБ}} = 10 \lg \left( E_b / N_0 \right), ~\text{дБ}.$$

Если энергия привязана к безразмерным кодам символов модуляции, то и одностороннюю СПМ шума следует выражать в той же системе единиц, хотя изначально она измеряется в $[\text{Вт/Гц}]$.

При цифровой обработке смеси сигнала с шумом в согласованном фильтре (в приёмнике) величину односторонней СПМ шума на входе фильтра можно оценить по-разному в зависимости от выбранной нормировки. Используют две нормировки:

* По постоянному току (Unity Gain), когда сумма отсчётов импульсной характеристики (ИХ) фильтра равна единице, при этом коэффициент передачи фильтра по модулю равен единице на частоте максимума; для фильтров нижних частот (baseband) это нулевая частота,
* По энергии (Unit Energy), когда сумма квадратов отсчётов ИХ фильтра равна единице, при этом дисперсии шума на входе и выходе такого фильтра равны друг другу.

Для случая нормировки по постоянному току имеем:

$$N_0 = 2 \sigma^2 T.$$

Здесь $\sigma^2$ — дисперсия шума на выходе согласованного фильтра, $T$ — интервал следования символов.

Для случая нормировки по энергии имеем:

$$N_0 = 2 \sigma^2 T / N_s.$$

Здесь $N_s$ — количество отсчётов на символ модуляции (Samples per Symbol, SPS).

Величину дисперсии шума на практике можно оценить, например, выключив (т.е. положив равным нулю) полезный сигнал.

## Средняя мощность сигнала. Первичное определение.

Как уже было условлено, под сигналом будем понимать непрерывный сигнал, поэтому его можно обозначить как функцию времени, $s(t)$. Сигналы систем связи — это комплекснозначные функции времени. Напомним, что в системах связи величину сигнала (т.е. его амплитуду) принято считать безразмерной.

Мгновенной мощностью называется квадрат амплитуды сигнала, $p(t) = {\lvert s(t) \rvert}^2$. Тогда средней мощностью на интервале времени ${T}_{\text{изм}}$ будет величина:

$$P = 1/{T}_{\text{изм}} \int_{t_0}^{t_0 + {T}_{\text{изм}}} {p(t) dt}.$$

Здесь происходит интегрирование (непрерывное суммирование) мгновенных мощностей на некотором интервале времени (измерительном интервале) и деление результата на этот интервал. Моментом начала интегрирования является произвольно выбранный момент $t_0$. Напомним, что величина сигнала — безразмерная (нет привязки к току или напряжению), поэтому нет смысла связывать мощность с сопротивлением, как это принято в физике: достаточно учесть эффект формы сигнала, т.е. то, что мощность пропорциональна квадрату сигнала.

Для сигналов цифровых систем связи характерна своего рода стационарность, когда сигнал — если наблюдать десятки и сотни символов модуляции подряд — выглядит достаточно однообразным, поэтому есть основания предполагать стабилизацию средней мощности $P$ при увеличении интервала измерения ${T}_{\text{изм}}$, а также её независимость от момента начала измерения.

## Средняя мощность цифрового сигнала.

Да, это не опечатка. Снова про среднюю мощность, но уже с другой — более определённой стороны.

Введённая выше средняя мощность обладает тем недостатком, что она является приближённой, так как зависит от двух произвольных параметров $t_0$ и ${T}_{\text{изм}}$.

Для систем связи можно выделить характерный момент времени — интервал следования символов модуляции, $T$, который определяет техническую скорость передачи информации (в бодах). Поэтому появляется возможность задать интервал интегрирования равным (или кратным) этому символьному интервалу. Однако при фиксации интервала наблюдения точность измерения становится неконтролируемой, поэтому усреднение делают статистически, что показано линией сверху интеграла:

$$P = {{1} \over {kT}} \overline{\int_{t_0}^{t_0 + kT} {p(t) dt}}.$$

Под статистическим усреднением в контексте цифровых систем связи понимают усреднение по всем информационным импульсам, которые соответствуют символам модуляции:

$$P = {{1} \over {kT}} \sum_{i=0}^{M-1} \int_{t_0}^{t_0 + kT} {p_i {\lvert s_i(t) \rvert}^2 dt}.$$

Здесь $p_i$ — вероятность $i$-го символа модуляции, которому соответствует импульс $s_i(t)$.

Обычно все символы модуляции равновероятны. Параметр $k = 1, 2, \ldots$ определяется длительностью информационного импульса, который был выбран для сигнального представления символов модуляции. Импульс сам по себе может занимать несколько символьных интервалов — это особенно актуально для систем радиосвязи, где важна компактность спектра излучаемого радиосигнала.

Определённая таким образом мощность является наиболее точной.

## Средняя мощность сигнала по сигнальному созвездию.

И, наконец, последняя ремарка про среднюю мощность. На этот раз с самой простой и практичной стороны.

Недостатком определённой выше средней мощности является сложность расчётной формулы и зависимость от внешнего параметра $k$, который в большей степени определяет компактность спектра формируемого сигнала, чем среднюю мощность.

Чтобы упростить формулу расчёта средней мощности, интеграл заменяют конкретным значением, а суммирование оставляют неизменным:

$$P = \sum_{i=0}^{M-1} {p_i {\lvert s_i \rvert}^2}.$$

Конкретное значение ${\lvert s_i \rvert} \neq 0$ выбирают равным кодовому числу, которое соответствует $i$-му символу модуляции. Таким числом является комплексное число, отображаемое точкой на комплексной плоскости; набор таких точек называют сигнальным созвездием, которое и задаёт вид модуляции.

## Цифровая модуляция (манипуляция). Модель цифрового сигнала.

Процесс преобразования символа модуляции $s_i$ в некоторый сигнал $s_i(t)$ называется цифровой модуляцией (манипуляцией). Обратный процесс, соответственно, называется демодуляцией. Модуляция соответствует передающей стороне (передатчику), а демодуляция — приёмной стороне (приёмнику).

Из-за неидеальности канала связи процесс демодуляции сопровождается ошибками. Качество демодуляции определяется вероятностью символьной ошибки (SER), которая зависит от отношения сигнал-шум в приёмнике. Каждый вид модуляции имеет свою зависимость вероятности символьной ошибки от отношения сигнал-шум. Вероятность символьной ошибки можно — точно или приближённо — пересчитать в вероятность ошибки на один бит для целей сравнения систем связи друг с другом.

Напомним понятие радиосигнала. Радиосигнал — это сигнал с центральной частотой $f_0$, которая намного больше ширины спектра сигнала. Такой сигнал математически можно записать в виде следующего достаточно общего выражения:

$$s(t) = A(t) e^{j \theta(t)} = A(t) e^{j \varphi(t)} e^{j 2 \pi f_0 t}.$$

Здесь $A(t)$ — медленно изменяющаяся амплитуда, $\varphi(t)$ — медленно изменяющаяся фаза, $\theta(t)$ — полная фаза сигнала.

Медленно изменяющиеся величины — это информационные величины (именно в них заложена информация). Их медленность обеспечивается условием малости ширины спектра радиосигнала по сравнению с центральной частотой.

Модуляция (манипуляция) бывает:

* Линейной, когда информационный импульс $g(t)$ линейно модулирует сигнал $s(t)$:

$$v_I(t) = \sum_{n \in \mathbb{Z}} {I_n g(t - nT)},$$

$$v_Q(t) = \sum_{n \in \mathbb{Z}} {Q_n g(t - nT)},$$

$$s(t) = \left[v_I(t) + j v_Q(t)\right] e^{j 2 \pi f_0 t}.$$

* Нелинейной, когда информационный импульс $g(t)$ модулирует сигнал $s(t)$ через некоторое нелинейное преобразование $F(v(t))$, например:

$$s(t) = A e^{j \theta(t)} = A e^{j 2 \pi f_0 t + K \int_{-\infty}^{t} {j v(\tau)} d\tau}.$$

Буквами $I$ и $Q$ обозначаются квадратурные компоненты сигнала $s(t)$. Напомним, что если у радиосигнала обнулить центральную частоту $f_0 \rightarrow 0$, то останется комплексная амплитуда:

$$\dot{A}(t) = A(t) e^{j \varphi(t)} =$$
$$ = A(t) \cos(\varphi(t)) + j A(t) \sin(\varphi(t)) =$$
$$ = I(t) + j Q(t),$$

которая состоит из квадратур — вещественной $I(t)$ и мнимой $Q(t)$ частей.

Для случая нелинейной модуляции приведён пример частотной манипуляции (ЧМн), когда мгновенная частота $f(t)$ сигнала $s(t)$ управляется сигналом $v(t)$:

$$f(t) := \frac{1}{2\pi} \frac{d\theta}{dt} =$$

$$= \frac{1}{2\pi} \frac{d}{dt} \left[ 2\pi f_0 t + K \int_{-\infty}^{t} v(\tau) d\tau \right] =$$

$$= f_0 + \frac{K}{2\pi} v(t).$$

Здесь управляющий сигнал $v(t)$ состоит из одной квадратуры, хотя после модуляции сигнал будет содержать две квадратуры.

Под $I_n$ и $Q_n$ понимаются две последовательности символов, образующие комплексный цифровой поток в дискретном времени $n \in \mathbb{Z}$. Например, для квадратурной амплитудной манипуляции (КАМ-М) характерно то, что квадратурные компоненты $I_n$ и $Q_n$ принимают нечётные значения из диапазона целых чисел $\pm 1, \pm 3, \ldots$, хотя в общем случае цифровая последовательность принимает произвольные вещественные значения (кроме разве что нуля), зависящие от конкретной схемы модуляции.

Под $f_0$ понимают некоторую несущую частоту (или промежуточную частоту), которой в общем случае может и не быть, как, например, в проводных системах связи с медным кабелем.

Случай линейной модуляции задаёт модель цифрового сигнала для большинства цифровых модуляций:

* Амплитудной (АМн),
* Фазовой (ФМн), 
* Амплитудно-фазовой (АФМн) и 
* Квадратурной амплитудной манипуляции (КАМ-М).

## Информационные импульсы. Тактовая синхронизация.

Перед тем как передавать цифровую информацию через канал (линию) связи, требуется определиться с сигналом, который будет переносить информацию.

В первом приближении требуемая ширина канала определяется интервалом следования символов модуляции, $T$:

$$\Delta f \sim 1/T.$$

Однако длительность ${T}_{\text{pulse}}$ информационного импульса $g(t)$

$${T}_{\text{pulse}} = kT, \quad k = 1, 2, \ldots,$$

а также его форма более тонко, но всё же достаточно существенно определяют требования к каналу, потому что качество систем связи определяется в том числе системой тактовой синхронизации, которая восстанавливает интервал следования символов модуляции с точностью до фазы. Слишком узкий канал приведёт к недопустимому искажению формы сигнала и сбою системы синхронизации, потому что система тактовой синхронизации настраивается исходя из особенностей формы сигнала, которая полностью определяется формирующим фильтром.

## Формирующий фильтр. Межсимвольная интерференция.

Чем компактнее должен быть спектр формируемого цифрового сигнала, тем больше должна быть длительность информационного импульса. Это можно наглядно подкрепить той аналогией, что самый узкий по спектру сигнал — «чистый тон» (т.е. синус) — имеет бесконечную длительность. Практически же используют функции типа синуса Котельникова с разными оконными функциями (например, с фильтром «приподнятого косинуса»). При этом, чтобы исключить взаимное влияние импульсов в моменты приёма, от информационных импульсов $g(t)$ в простейшем случае требуется соблюдение отсчётного свойства:

$$g(nT) = 0, ~n \neq 0,$$

$$g(nT) = 1, ~n = 0.$$

Таким образом, в идеальном случае (при отсутствии шумов и при точной работе системы тактовой синхронизации) в моменты времени, кратные интервалу следования символов $T$, исключается межсимвольная интерференция (МСИ).

Ниже проиллюстрировано отсчётное свойство фильтра «приподнятого косинуса» на примере приёма четырёх двоичных символов $[+1, +1, -1, -1]$.

```python
import matplotlib.pyplot as plt
import numpy as np
import ipywidgets as widgets
from IPython.display import display

def raised_cosine(t, T, beta):
    """Функция приподнятого косинуса."""
    t = np.asarray(t, dtype=float)
    if beta == 0:
        return np.sinc(t / T)
    with np.errstate(divide='ignore', invalid='ignore'):
        # numpy.sinc(x) — это sin(pi*x)/(pi*x)
        sinc = np.sinc(t / T)
        cos_part = np.cos(np.pi * beta * t / T)
        denom = 1 - (2 * beta * t / T)**2
        res = sinc * cos_part / denom
        # Обработка точек, где знаменатель равен 0
        limit = (np.pi / 4) * np.sinc(1 / (2 * beta))
        res = np.where(np.abs(denom) < 1e-10, limit, res)
    return res

# 1. Создаём виджет
float_slider = widgets.FloatSlider(
    value=0.5,           # Начальное значение
    min=0,               # Минимум
    max=1.0,            # Максимум
    step=0.05,            # Шаг изменения
    description='Коэффициент скругления:',
    # 'initial' заставит виджет подстроиться под длину текста
    style={'description_width': 'initial'},
    readout_format='.2f' # Формат отображения (2 знака после запятой)
)

# 2. Функция отрисовки
def plot_func(beta):
    # Параметры
    T = 1.0          # Символьный интервал
    sps = 12         # Отсчётов на символ
    span = 8        # Увеличим охват символов для наглядности
    
    # Временная ось (расширена для четырёх импульсов)
    t = np.linspace(-span*T/2, span*T/2 + T, span*sps + sps + 1)
    
    # Генерация четырёх импульсов с шагом T
    p1 = 1*raised_cosine(t + T, T, beta) # Центрирован в -T
    p2 = 1*raised_cosine(t, T, beta)     # Центрирован в 0
    p3 = -1*raised_cosine(t - T, T, beta) # Центрирован в +T
    p4 = -1*raised_cosine(t - 2*T, T, beta) # Центрирован в +2T
    
    # Суммарный сигнал
    v = p1 + p2 + p3 + p4
    
    # Визуализация
    plt.figure(figsize=(10, 6))
    
    plt.plot(t, p1, 'g--', label='Импульс +g(t+T)', alpha=0.5)
    plt.plot(t, p2, 'b--',  label='Импульс +g(t)', alpha=0.5)
    plt.plot(t, p3, 'r--', label='Импульс -g(t-T)', alpha=0.5)
    plt.plot(t, p4, 'm--', label='Импульс -g(t-2T)', alpha=0.5)
    plt.plot(t, v, 'k', label='Суммарный сигнал v(t)', linewidth=1.7)
    
    # Моменты дискретизации
    sampling_instants = [-T, 0, T, 2*T]
    plt.scatter(sampling_instants, [1, 1, -1, -1], color='red', zorder=5, label='Точки отсчёта')
    
    # Линии для наглядности
    for ts in sampling_instants:
        plt.axvline(ts, color='gray', linestyle=':', alpha=0.4)
    
    plt.xticks(fontsize=16)
    plt.yticks(fontsize=16)
    
    plt.title('Приём последовательности символов [+1, +1, -1, -1]', fontsize=19)
    plt.xlabel('Время (t/T)', fontsize=16)
    plt.ylabel('Амплитуда', fontsize=16)
    plt.legend(loc='upper right')
    plt.grid(True, alpha=0.3)
    plt.axhline(0, color='black', linewidth=0.8)
    plt.ylim(-1.5, 1.5)
    
    plt.show()

# 3. Связываем виджет и функцию
# Ключ в словаре должен совпадать с названием аргумента функции
out = widgets.interactive_output(plot_func, {'beta': float_slider})

# 4. Отображаем всё вместе
display(float_slider, out)
```

Фильтр приподнятого косинуса является фильтром без МСИ. Однако такая картина должна быть в точке приёма, а не на стороне передатчика, потому что принятие решения о переданном символе происходит в приёмнике. По этой причине фильтр без МСИ (любой, который тем или иным образом найден, не обязательно «приподнятый косинус») расщепляют на два идентичных фильтра так, что их каскадное соединение даёт фильтр без МСИ. Каскадное соединение — это естественное последовательное соединение фильтров в тракте «передатчик — приёмник». В связи с этим в реальной системе связи на выходе передатчика (на выходе формирующего фильтра) МСИ присутствует, однако это присутствие создано намеренно, так что обработка сигнала в приёмнике в идеальных условиях даст отсутствие МСИ в полном соответствии с проектными расчётами.

## Согласованный фильтр. Оптимальный приём.

Так как система связи предполагает две неотъемлемые части: передатчик (Tx) и приёмник (Rx), то исходный фильтр без МСИ расщепляют на два идентичных фильтра: формирующий (передающий) и согласованный (приёмный):

$$H(f) = H_\text{Tx}(f) H_\text{Rx}(f), \quad H_\text{Tx}(f) = H_\text{Rx}(f) = \sqrt{H(f)}.$$

Это позволяет в приёмнике естественным образом выполнять оптимальный приём, получая при этом наибольшее отношение сигнал-шум после обработки:

$$q_\text{max}^2 = {{2E_d} \over {N_0}}.$$

Здесь под $E_d$ имеется в виду энергия разностного сигнала. Под разностным сигналом понимается разность двух эталонных сигналов, амплитуда которых приведена к некоторому номинальному уровню (считается, что уровни поддерживаются системой автоматической регулировки усиления, АРУ):

$$s_d(t) = s_i(t) - s_j(t).$$

Здесь $i$ и $j$ — индексы двух ветвей в момент принятия решения. Для двоичных модуляций существуют всего два импульса, поэтому там решение принимается единожды. Для модуляций более высоких порядков для каждого бита выделяют виртуальные точки, относительно которых принимают бинарное решение. Часто используют код Грея: соседние точки созвездия отличаются только в одном бите; это снижает вероятность битовых ошибок и упрощает расчёт логарифмического отношения правдоподобия (LLR). Энергия такого виртуального разностного сигнала в общем случае зависит от позиции бита в символе модуляции.

Примером расщеплённого фильтра является фильтр типа «корень из приподнятого косинуса».

## Количество информации как мера снятия неопределённости при приёме. Максимальное правдоподобие. Вероятность ошибки.

Далее в нескольких параграфах будут рассмотрены аспекты теории информации.

Предупреждение: ко всему, что связано с применимостью теории информации к системам связи, следует относиться внимательно, т.к. данная область, несмотря на простоту формул, достаточно коварна в отношении смысла описываемых величин.

### Собственная информация

Для раскрытия данной темы достаточно рассмотреть модель системы передачи информации в виде источника с алфавитом $X$, канала связи и приёмника.

Напомним, что понятие канала связи растяжимо, и, соответственно, в зависимости от уровня рассмотрения для одной и той же системы связи могут быть разные пары алфавитов $X$ и $Y$:

$$X \to X' \to X'' \to \ldots \to Y'' \to Y' \to Y.$$

Фактически, данное выражение иллюстрирует вложенность каналов или, лучше сказать, многоступенчатость процесса формирования (передатчик) и обработки (приёмник) сигнала, если для упрощения под сигналом понимать нечто общее, в которое включено и кодирование. Многоступенчатость при разумном её внедрении позволяет повысить качество системы связи в целом. Однако чрезмерная многоступенчатость вредит (в лучшем случае приводит к лишнему расходу энергии). Выбранную конкретную пару $(X, Y)$ можно называть интерфейсом канала связи. Здесь $X$ — алфавит на входе канала, $Y$ — алфавит на выходе. Например, можно рассматривать двоичный канал со стираниями, тогда $X$ — двоичный алфавит $\lbrace 0, 1 \rbrace$, а $Y$ — троичный алфавит $\lbrace 0, 1, \text{e} \rbrace$, где $\text{e}$ — стёртый символ. 

Приёмник выступает в роли наблюдателя, и по результату наблюдения $y_j$ делает выводы о текущем значении $x_i$ символа источника. Выводы делаются на основе принципа максимального правдоподобия. Однако чтобы это правдоподобие посчитать, требуется ввести в рассмотрение отношение вероятностей.

Рассмотрим в качестве примера процесс подбрасывания шестигранного кубика и показ выпавшей цифры стороннему наблюдателю. В этом случае размер алфавита $X$ равен $6$, и символ может с равной вероятностью принять одно из шести значений $x_i$, $0 \leq i < 6$. До подбрасывания кубика у наблюдателя имеется неопределённость относительно ожидаемых значений. В общей теории связи (по Клоду Шеннону) эта неопределённость количественно выражается через двоичный логарифм обратной вероятности выпадения конкретного значения:

$$H(x_i) = \log_2 {{1} \over {p(x_i)}} \geq 0, ~\text{бит}.$$

Если бы вероятность выпадения конкретного значения символа была равна единице, то не было бы смысла передавать такой символ (осуществлять опыт подбрасывания). Источник $X$ был бы в этом случае детерминированным, а неопределённость была бы равна нулю. Смысл передачи и приёма возникает только при ненулевой величине неопределённости, потому что приём — это процесс снятия неопределённости (в идеале до нуля). Считается, что полное снятие неопределённости — это теоретический предел, который на практике недостижим.

Неопределённость измеряется в битах, потому что основание логарифма было выбрано равным $2$. Естественно, для внутренних расчётов основания можно пересчитывать друг в друга, но для единообразия принято использовать единицу измерения $[\text{бит}]$. Бит — это единица измерения информации, соответствующая одному двоичному символу, то есть выбору из минимального нетривиального алфавита из двух значений.

В рассматриваемом примере наблюдатель идеальный, и потому неопределённость снимается до нуля, а величину $H(x_i)$ называют собственной информацией значения $x_i$. Обычно её обозначают как $I(x_i) = H(x_i)$. То, что наблюдатель идеальный, подчёркивает единица в числителе, которая есть не что иное, как единичная вероятность. После опыта (т.е. наблюдения конкретного значения) вероятность этого значения становится равной единице, $p(x_i) \rightarrow 1$, и потому неопределённость снимается (до нуля) на величину $H(x_i)$. Это равнозначно получению информации $I(x_i)$.

### Взаимная информация

В общем и наиболее интересном случае наблюдатель неидеальный, и в канале связи есть помехи. Тогда количество получаемой информации определяется следующей формулой:

$$I(x_i, y_j) = {\log_2 {{p(x_i \mid y_j)} \over {p(x_i)}}}, \text{бит}.$$

Данную величину называют взаимной информацией между значениями $x_i$ и $y_j$. Вероятность $p(x_i \mid y_j)$ — это условная вероятность; вертикальная черта отделяет событие от условия. В данном случае $p(x_i \mid y_j)$ — вероятность значения $x_i$ при условии, что наблюдается значение $y_j$. Величина $I(x_i, y_j)$ может принимать любое значение, в том числе отрицательное; средняя взаимная информация $I(X, Y)$ неотрицательна.

Величина $I(x_i, y_j)$ определяет, сколько информации будет получено о значении $x_i$ при наблюдении $y_j$.

Величина $I(x_i, y_j)$ показывает, насколько наблюдение $y_j$ изменяет вероятность $x_i$ (для каждого $i$). Приёмник принимает решение о $x_i$ по максимуму правдоподобия. Значение $y_j$ на практике — это некоторое число (как правило, ограниченной разрядности — мягкий приём), по которому делается вывод о наиболее правдоподобном значении $x_i$ символа $X$. После того как вывод сделан, мы, условно говоря, приходим на самый конец приёмника, $X \rightarrow Y \rightarrow Z$, где $Z$ — алфавит решений приёмника, совпадающий с $X$. Ошибка определяется по решению: при наличии ошибки индексы у символов $z_j$ и $x_i$ не равны, $j \neq i$, а при отсутствии ошибки $j = i$. Для каналов связи без помех (недостижимый идеал) всегда $p(z_i \mid x_i) = 1$, а $p(z_j \mid x_i) = 0$. В канале связи с помехами $p(z_i \mid x_i) < 1$, а для некоторых $j \neq i$ вероятность $p(z_j \mid x_i) > 0$; величина $p(z_j \mid x_i)$ называется вероятностью ошибки $i \rightarrow j$. Естественно, в общем случае возможны разные комбинации ошибок, и каждая из них может иметь разную вероятность.

Здесь символ $Y$ характеризует промежуточный этап, который был рассмотрен для иллюстрации сути наблюдения в реальных приёмниках, при этом под приёмником понимается согласованный приём плюс детектирование символа — сейчас это просто алгоритмы, реализованные на чипах. Для символа $Z$, однако, также можно вычислить взаимную информацию (относительно $X$), правда, наблюдения $Z$ как такового уже нет, потому что это конечный результат детектирования. Зная природу наблюдения, можно вычислить вероятности $p(x_i \mid z_j)$.

## Производительность источника (информационная энтропия). Скорость передачи информации. Требуемая избыточность кода.

Так как в канале с помехами возможны разные комбинации ошибок $i \rightarrow j$, вводят усреднение значений информации — как собственной, так и взаимной — по всем возможным комбинациям. Это позволяет получить средние величины, которые называются, соответственно, производительностью источника (или его информационной энтропией) $H(X)$ 

$$H(X) = \sum_{i} {p(x_i) I(x_i)}, \text{бит/символ},$$

и скоростью передачи информации (средней взаимной информацией) $I(X, Y)$:

$$I(X, Y) = \sum_{i} \sum_{j} {p(x_i, y_j) I(x_i, y_j)}, \text{бит/символ}.$$

Здесь $p(x_i, y_j)$ — совместная вероятность появления значений $x_i$ и $y_j$. Размеры алфавитов намеренно не конкретизируются, а суммы подразумевают суммирование по всем допустимым индексам.

Если приведённые здесь средние величины разделить на интервал следования $T$ (в секундах) символов, то их единица измерения будет равна $[\text{бит/с}]$, и неважно, какой размер алфавита, потому что перевод в биты уже сделан выбором основания логарифма.

Известны следующие свойства средних величин:

$$0 \leq H(X) \leq \log_2 M,$$

$$0 \leq I(X, Y) \leq H(X).$$

Здесь $M$ — размер алфавита. Производительность источника достигает максимума при равновероятных символах, а ненулевая разница

$$H(X \mid Y) = H(X) - I(X, Y), \text{бит/символ},$$

вызвана влиянием помех в канале связи. Эту разницу называют потерями в канале (условной энтропией), и именно её компенсируют с помощью корректирующих кодов с избыточностью $0 \leq {R}_{\text{ex}} < 1$. Чтобы сравнивать избыточность конкретного кода с условной энтропией, последнюю следует нормировать, т.е. поделить на $\log_2 M$:

$${R}_{\text{ex}} > {{H(X \mid Y)} \over {\log_2 M}}.$$

Это условие является теоретическим: при его выполнении существует код, обеспечивающий сколь угодно малую вероятность ошибки при достаточно большой длине кодового слова. Оно не гарантирует исправления всех ошибок конкретным кодом конечной длины. Практически, избыточность требуется взять с некоторым запасом, и эта область больше связана с практикой, нежели с теоретическими рассуждениями (ситуация будет зависеть от конкретного кода и конкретного алгоритма декодирования). Однако код с избыточностью ниже указанной не позволит обеспечить сколь угодно малую вероятность ошибки при данном распределении источника. Именно этим и ценна теория Клода Шеннона.

Заметим, что можно исключить множитель $\log_2 M$, понимая под символами всегда двоичные кодовые слова (теоретически бит — единица информации, соответствующая двоичному символу):

$$M = 2^n.$$

Здесь $n$ — длина кодового слова. Конечно, это ограничивает диапазон $M$, однако теория Шеннона даёт предельные соотношения, и поэтому можно считать $M$ очень большим, так что это и есть некоторая степень двойки. Такое исключение множителя позволяет автоматически совершить предельный переход к бесконечной последовательности битов $n \rightarrow \infty$, что в реальности и происходит — количество битов в современных системах связи огромное. Тогда:

$${R}_{\text{ex}} > {H(X \mid Y) \over n}.$$

Если теперь под $H(X \mid Y) / n$ при $n \rightarrow \infty$ понимать удельную условную энтропию (удельные потери в канале)

$$H_{\text{уд}}(X \mid Y) = \lim_{n \to \infty} H(X \mid Y) / n, ~\text{бит/бит}$$

то формула сведётся к:

$${R}_{\text{ex}} > H_{\text{уд}}(X \mid Y).$$

В действительности же никаких предельных переходов совершать не требуется: достаточно взять битовый уровень рассмотрения канала выбранной системы связи (какими бы ни были модуляция и кодирование), и тогда автоматически условная энтропия будет равна удельной, что позволит сравнивать её с избыточностью кода.

Размерность $[\text{бит/бит}]$ для удельной условной энтропии подразумевает, сколько в среднем бит информации теряется в канале в расчёте на один физически переданный (и принятый) бит.

## Пропускная способность канала связи. Мягкий и жёсткий приём.

Пропускная способность канала — это максимальная скорость передачи информации, при которой ещё можно обеспечить сколь угодно малую вероятность ошибки. Это предельная характеристика самого канала, а не скорость конкретной системы связи.

### Непрерывный канал

Для непрерывного канала с аддитивным гауссовским шумом справедлива формула Шеннона:

$$C = \Delta f_{\text{ш}} \log_2 \left(1 + S/N\right), \text{бит/с}.$$

Здесь $\Delta f_{\text{ш}}$ — односторонняя шумовая полоса частот, а $S/N$ — отношение средней мощности сигнала к мощности шума. Формула показывает, что пропускную способность можно повышать, расширяя полосу или увеличивая мощность сигнала. При этом рост от мощности получается логарифмическим: каждый дополнительный бит в секунду требует всё большего превышения сигнала над шумом.

### Дискретный канал

Для цифрового канала удобнее не работать с непрерывными сигналами напрямую, а описать канал через вероятности переходов. В цифровой связи рассматривают дискретный канал без памяти, в котором переходные вероятности не зависят от предыдущих символов. Зададим алфавиты:

* $X = \{x_0, \ldots, x_{M-1}\}$ — алфавит символов на входе канала;
* $Y = \{y_0, \ldots, y_{N-1}\}$ — алфавит символов на выходе канала.

Канал полностью описывается переходными вероятностями $p(y_j \mid x_i)$: с какой вероятностью при передаче $x_i$ будет принято $y_j$. Чем выше качество канала, тем ближе эти вероятности к единице при $j = i$ и к нулю при $j \neq i$. Улучшение отношения сигнал-шум и корректирующие коды уменьшают вероятности ошибок, приближая канал к идеальному.

Если известны вероятности входных символов $p(x_i)$, можно вычислить среднюю взаимную информацию $I(X,Y)$. Она показывает, сколько информации в среднем переносится одним символом канала. Пропускная способность — это максимум $I(X,Y)$ по всем допустимым распределениям источника:

$$C_M = \max_{p(X)} I(X,Y), \text{бит/символ}.$$

Максимизация нужна потому, что при одном и том же канале разные распределения источника дают разную скорость. Распределение, при котором достигается максимум, называют согласованным с каналом. Для симметричных каналов оно часто оказывается равномерным.

### Мягкий и жёсткий приём

В этой модели выходной алфавит $Y$ не обязан совпадать с входным. Это и объясняет различие между мягким и жёстким приёмом.

При жёстком приёме приёмник сначала принимает решение о переданном символе. На выходе получается тот же алфавит, что и на входе: $Y = X$. При мягком приёме приёмник сохраняет больше информации: например, квантованные отсчёты сигнала или значения логарифмического отношения правдоподобия. Выходной алфавит $Y$ при этом может содержать гораздо больше уровней, а в пределе — быть непрерывным.

Чем больше информации сохраняет приёмник, тем выше взаимная информация $I(X,Y)$ и пропускная способность такого канала. Поэтому мягкий приём потенциально эффективнее жёсткого. На практике степень «мягкости» можно регулировать — вплоть до жёсткого решения. В пределе, когда выходной алфавит становится непрерывным, дискретный канал переходит в аналоговый.

### Нормировка на двоичный символ

Величина $C_M$ зависит от размера входного алфавита $M$, поэтому неудобна для сравнения разных систем. Чтобы получить универсальную характеристику, пропускную способность нормируют на один двоичный символ. Для двоичного входа ($M = 2$):

$$C := C_2 = \max_{p(X)} I(X,Y), \text{бит/бит}.$$

Так как один двоичный символ не может переносить больше одного бита информации, $C \leq 1$. Величина $C$ показывает, сколько бит информации в среднем приходится на один физический бит, переданный по каналу.

### Что это даёт на практике

Если скорость кодирования $R$ меньше пропускной способности данного канала, то при достаточно большой длине кодового слова существует код, обеспечивающий сколь угодно малую вероятность ошибки:

$$R < C.$$

Через избыточность $R_{\text{ex}} = 1 - R$ это условие записывается так:

$$R_{\text{ex}} > 1 - C.$$

Если пропускная способность известна, предельную битовую скорость можно оценить по формуле

$$V = \frac{C N_b}{T}, \text{бит/с},$$

где $N_b$ — количество физических битов, переданных за время $T$. Большой интервал $T$ выбирают, чтобы учесть служебную структуру системы: кадры, фреймы, многоканальность и т.п.

Условия приёма в реальном канале меняются, поэтому меняются переходные вероятности и пропускная способность. Адаптивные системы связи подстраивают модуляцию и кодирование, чтобы поддерживать качество связи и пропускную способность на стабильном уровне.


## Спектральная эффективность систем связи.

Ширина канала (односторонняя) при использовании формирующего фильтра типа «приподнятый косинус» равна:

$$\Delta f = {{1 + \beta} \over {2T}}.$$

Здесь $0 \leq \beta \leq 1$ — коэффициент скругления (roll-off factor), определяемый формирующим фильтром.

При передаче сигнала на некоторой несущей ширина спектра возрастает в два раза; используя эту ширину и битовую скорость передачи, вводят понятие спектральной эффективности:

$$\eta_s = V_b / 2\Delta f = {{R \log_2 M} \over {1 + \beta}}, ~\text{бит/с/Гц}.$$

Для систем мобильной связи типовые значения спектральной эффективности лежат в диапазоне $0{,}1 < \eta_s < 10$ в зависимости от стандарта: 2G, 3G, 4G и 5G. Переход через единицу произошёл на стыке сетей третьего и четвёртого поколений. Следует отметить, что эффективность систем связи — статистическая величина, т.к. современные системы связи адаптируются к текущим условиям приёма, переключая кодирование и модуляцию для поддержания заданной вероятности битовой ошибки.

## Энергетическая эффективность систем связи.

Данная величина дополняет рассмотренную выше спектральную эффективность:

$$\eta_e = V_b / P, ~\text{бит/Дж}.$$

Здесь $P$ — средняя потребляемая системой мощность.

В контексте общей теории связи под средней мощностью понимают среднюю мощность, определяемую по сигнальному созвездию. Для квадратурных созвездий (например, КАМ) такая средняя мощность в простейшей оценке пропорциональна порядку модуляции, т.е. количеству символов созвездия:

$$P \sim M d_{\text{min}}^2.$$

Здесь ${d}_{\text{min}}$ — минимальное расстояние между точками сигнального созвездия. Для других типов созвездий, например, ФМн, зависимость может быть иной. Рост порядка модуляции улучшает спектральную эффективность, но ухудшает энергетическую.


## Ресурсная эффективность систем связи.

Взвешенное произведение спектральной и энергетической эффективностей даёт ресурсную эффективность системы связи:

$$\eta_r = {\eta_s}^{w} {\eta_e}^{1 - w}.$$

В зависимости от типа системы связи отдают приоритет либо спектральной эффективности, либо энергетической, регулируя параметр $0 \leq w \leq 1$. Современные мобильные системы связи (5G и 6G) могут регулировать этот параметр, например, в зависимости от времени суток, отдавая днём приоритет спектральной эффективности, а ночью — энергетической.

Иногда говорят об энергоспектральной эффективности, когда при проектировании находят «точку равновесия», при которой система не тратит лишнюю энергию ради незначительного прироста скорости.
